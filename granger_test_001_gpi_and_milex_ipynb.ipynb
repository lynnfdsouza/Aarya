```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import chi2_contingency
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import uuid

# Define NATO countries (as of 2025)
nato_countries = [
    'Albania', 'Belgium', 'Bulgaria', 'Canada', 'Croatia', 'Czech Republic', 'Denmark', 'Estonia',
    ' Finland', 'France', 'Germany', 'Greece', 'Hungary', 'Iceland', 'Italy', 'Latvia', 'Lithuania',
    'Luxembourg', 'Netherlands', 'Norway', 'Poland', 'Portugal', 'Romania', 'Slovakia', 'Slovenia',
    'Spain', 'Sweden', 'Turkey', 'United Kingdom', 'United States'
]


# Merge datasets - using the dataframes loaded from the excel files
# Renaming year columns before merging to avoid conflicts and have meaningful suffixes
gpi_years = {year: f'{year}_y' for year in range(2014, 2024)}
sipri_years = {year: f'{year}_x' for year in range(2014, 2024)}

gpi_df_renamed = gpi_df.rename(columns=gpi_years)
sipri_df_renamed = sipri_df.rename(columns=sipri_years)


merged_df = pd.merge(sipri_df_renamed, gpi_df_renamed, on='Country', how='inner')

# Add NATO membership column
merged_df['NATO'] = merged_df['Country'].apply(lambda x: 'NATO' if x in nato_countries else 'Non-NATO')

# Define conflict status (high conflict if Ongoing Conflict score > 2)
conflict_cols = [f'{year}_y' for year in range(2014, 2024)]
merged_df['Conflict'] = merged_df[conflict_cols].gt(2).any(axis=1)

# Fill missing expenditure values with median
expenditure_cols = [f'{year}_x' for year in range(2014, 2024)]
merged_df[expenditure_cols] = merged_df[expenditure_cols].fillna(merged_df[expenditure_cols].median())

# 1. Correlation Heatmap
all_years = range(2014, 2024)
expenditure_cols_merged = [f'{year}_x' for year in all_years]
conflict_cols_merged = [f'{year}_y' for year in all_years]
corr_data = merged_df[expenditure_cols_merged + conflict_cols_merged]
corr = corr_data.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap: Military Expenditure vs Ongoing Conflict (2014-2023)')
plt.tight_layout()
plt.show()

# 2. Chi-Square Test
contingency_table = pd.crosstab(merged_df['NATO'], merged_df['Conflict'])
chi2, p, dof, expected = chi2_contingency(contingency_table)
print(f"Chi-Square Test Results:\nChi2 Statistic: {chi2:.2f}\nP-value: {p:.4f}\nDegrees of Freedom: {dof}")

# 3. K-Means Clustering
# Prepare data (2023 expenditure and conflict score)
X = merged_df[['2023_x', '2023_y']].values  # Expenditure and Conflict score for 2023
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply K-Means
kmeans = KMeans(n_clusters=3, random_state=42, n_init='auto')
merged_df['Cluster'] = kmeans.fit_predict(X_scaled)

# Visualize clusters
plt.figure(figsize=(10, 6))
sns.scatterplot(x='2023_x', y='2023_y', hue='Cluster', style='NATO', size='Conflict', data=merged_df)
plt.title('K-Means Clustering: Military Expenditure vs Conflict Score (2023)')
plt.xlabel('Military Expenditure 2023 (Constant US$ m)')
plt.ylabel('Ongoing Conflict Score 2023')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

# Print cluster assignments
print("\nCluster Assignments:")
for cluster in sorted(merged_df['Cluster'].unique()):
    print(f"\nCluster {cluster}:")
    cluster_countries = merged_df[merged_df['Cluster'] == cluster][['Country', 'NATO', 'Conflict', '2023_x', '2023_y']]
    print(cluster_countries)
```

Correlation Heatmap: The heatmap shows the correlation coefficients between military expenditure and ongoing conflict scores across the years 2014-2023. A positive correlation indicates that as one variable increases, the other tends to increase as well. A negative correlation indicates that as one variable increases, the other tends to decrease. Coefficients close to 1 or -1 indicate a strong correlation, while coefficients close to 0 indicate a weak correlation. Based on the heatmap generated, you can observe the strength and direction of the relationship between military expenditure and ongoing conflict for each pair of years. It appears there are generally weak positive correlations between military expenditure and ongoing conflict scores across the years.

Chi-Square Test Results: The Chi-Square Test for independence examines whether there is a statistically significant association between two categorical variables. In this case, the test is assessing if there is a relationship between NATO membership (NATO/Non-NATO) and conflict status (True/False).

Chi2 Statistic (15.33): This is the value of the test statistic.
P-value (0.0001): This is the probability of observing a Chi-Square statistic as extreme as, or more extreme than, the one calculated, assuming there is no association between the two variables (the null hypothesis). A very small p-value (typically less than 0.05) suggests that the observed association is statistically significant, leading us to reject the null hypothesis. In this case, the p-value of 0.0001 is much less than 0.05, indicating a statistically significant association between NATO membership and conflict status in this dataset. This suggests that NATO membership and being in a conflict are not independent of each other based on this data.
Degrees of Freedom (1): This value is related to the number of categories in your variables and is used in determining the critical value for the Chi-Square test.
K-Means Clustering: The K-Means clustering grouped countries based on their 2023 military expenditure and ongoing conflict scores into 3 clusters.

Cluster 0: This cluster appears to contain a large number of countries with relatively lower military expenditure and generally lower ongoing conflict scores. Many of these countries are marked as 'Non-NATO' and 'False' for conflict, suggesting they are mostly peaceful and have lower military spending in 2023 compared to other clusters.
Cluster 1: This cluster seems to include countries with moderate to higher ongoing conflict scores and a range of military expenditures, generally higher than Cluster 0 but lower than Cluster 2. This cluster contains a mix of NATO and Non-NATO countries, and a significant number are marked as being in conflict ('True').
Cluster 2: This cluster appears to be dominated by the United States, which has a significantly higher military expenditure than any other country in the dataset, while having a relatively lower ongoing conflict score compared to countries in Cluster 1. This country stands out as an outlier in terms of military spending.

```python
# Calculate average military expenditure and ongoing conflict score per cluster per year
trends_df = merged_df.groupby('Cluster').agg({f'{year}_x': 'mean' for year in range(2014, 2024)} | {f'{year}_y': 'mean' for year in range(2014, 2024)}).T

# Reset index to have year as a column
trends_df = trends_df.reset_index()
trends_df = trends_df.rename(columns={'index': 'Year'})

# Reshape data for plotting
# The columns representing clusters are now integers 0, 1, and 2
cluster_cols = [0, 1, 2] # Assuming 3 clusters as in the previous K-Means

expenditure_trends_melted = trends_df[trends_df['Year'].str.contains('_x', na=False)].melt('Year', value_vars=cluster_cols, var_name='Cluster', value_name='Average Military Expenditure')
conflict_trends_melted = trends_df[trends_df['Year'].str.contains('_y', na=False)].melt('Year', value_vars=cluster_cols, var_name='Cluster', value_name='Average Ongoing Conflict Score')


# Convert year column to numeric for plotting
expenditure_trends_melted['Year'] = expenditure_trends_melted['Year'].str.replace('_x', '').astype(int)
conflict_trends_melted['Year'] = conflict_trends_melted['Year'].str.replace('_y', '').astype(int)

# Convert Cluster column to string for plotting
expenditure_trends_melted['Cluster'] = expenditure_trends_melted['Cluster'].astype(str)
conflict_trends_melted['Cluster'] = conflict_trends_melted['Cluster'].astype(str)


# Plot average military expenditure trends
plt.figure(figsize=(12, 6))
sns.lineplot(x='Year', y='Average Military Expenditure', hue='Cluster', data=expenditure_trends_melted)
plt.title('Average Military Expenditure Trends by Cluster (2014-2023)')
plt.xlabel('Year')
plt.ylabel('Average Military Expenditure (Constant US$ m)')
plt.xticks(range(2014, 2024))
plt.grid(True)
plt.show()

# Plot average ongoing conflict score trends
plt.figure(figsize=(12, 6))
sns.lineplot(x='Year', y='Average Ongoing Conflict Score', hue='Cluster', data=conflict_trends_melted)
plt.title('Average Ongoing Conflict Score Trends by Cluster (2014-2023)')
plt.xlabel('Year')
plt.ylabel('Average Ongoing Conflict Score')
plt.xticks(range(2014, 2024))
plt.grid(True)
plt.show()
```

Cluster 2 (dominated by the US) shows a significantly higher average military expenditure that remains relatively stable over the years, although with some fluctuation. Cluster 1 (high conflict countries) shows a lower average expenditure than Cluster 2, but with a noticeable upward trend, particularly in the later years. Analyzing Individual Country Deviations in Cluster 1:

```python
import statsmodels.api as sm

# Calculate individual country slopes in Cluster 1

selected_cluster = 1
cluster_df = merged_df[merged_df['Cluster'] == selected_cluster].copy() # Use a copy to avoid SettingWithCopyWarning

# Melt the dataframe for easier iteration by country and year
cluster_expenditure_melted = cluster_df.melt('Country', value_vars=[f'{year}_x' for year in range(2014, 2024)], var_name='Year', value_name='Military Expenditure')
cluster_conflict_melted = cluster_df.melt('Country', value_vars=[f'{year}_y' for year in range(2014, 2024)], var_name='Year', value_name='Ongoing Conflict Score')

# Convert year column to numeric
cluster_expenditure_melted['Year'] = cluster_expenditure_melted['Year'].str.replace('_x', '').astype(int)
cluster_conflict_melted['Year'] = cluster_conflict_melted['Year'].str.replace('_y', '').astype(int)

# Calculate average slopes for Cluster 1 (from previous analysis)
avg_expenditure_slope = expenditure_trends_melted[expenditure_trends_melted['Cluster'] == str(selected_cluster)].iloc[0]['Average Military Expenditure'] / expenditure_trends_melted['Year'].diff().dropna().mean()
avg_conflict_slope = conflict_trends_melted[conflict_trends_melted['Cluster'] == str(selected_cluster)].iloc[0]['Average Ongoing Conflict Score'] / conflict_trends_melted['Year'].diff().dropna().mean()

# Store individual country slopes
individual_expenditure_slopes = {}
individual_conflict_slopes = {}

# Calculate slopes for each country in Cluster 1
for country in cluster_df['Country'].unique():
    # Military Expenditure
    country_expenditure_data = cluster_expenditure_melted[cluster_expenditure_melted['Country'] == country].dropna()
    if len(country_expenditure_data) > 1:
        X = sm.add_constant(country_expenditure_data['Year'])
        y = country_expenditure_data['Military Expenditure']
        model = sm.OLS(y, X).fit()
        individual_expenditure_slopes[country] = model.params['Year']
    else:
        individual_expenditure_slopes[country] = np.nan # Not enough data points

    # Ongoing Conflict Score
    country_conflict_data = cluster_conflict_melted[cluster_conflict_melted['Country'] == country].dropna()
    if len(country_conflict_data) > 1:
        X = sm.add_constant(country_conflict_data['Year'])
        y = country_conflict_data['Ongoing Conflict Score']
        model = sm.OLS(y, X).fit()
        individual_conflict_slopes[country] = model.params['Year']
    else:
        individual_conflict_slopes[country] = np.nan # Not enough data points

# Convert slopes to pandas Series for easier comparison
individual_expenditure_slopes_series = pd.Series(individual_expenditure_slopes)
individual_conflict_slopes_series = pd.Series(individual_conflict_slopes)

# Calculate deviation from average slope
expenditure_slope_deviations = individual_expenditure_slopes_series - avg_expenditure_slope
conflict_slope_deviations = individual_conflict_slopes_series - avg_conflict_slope

# Display slopes and deviations
print(f"\nIndividual Country Military Expenditure Slopes (Cluster {selected_cluster}):")
display(individual_expenditure_slopes_series)
print(f"\nDeviation from Average Military Expenditure Slope (Cluster {selected_cluster}):")
display(expenditure_slope_deviations)

print(f"\nIndividual Country Ongoing Conflict Score Slopes (Cluster {selected_cluster}):")
display(individual_conflict_slopes_series)
print(f"\nDeviation from Average Ongoing Conflict Score Slopes (Cluster {selected_cluster}):")
display(conflict_slope_deviations)
```

```python
# Sort deviations to identify potential outliers

print("\nTop 5 Countries with Largest Positive Deviation in Military Expenditure Slope (Cluster 1):")
display(expenditure_slope_deviations.nlargest(5))

print("\nTop 5 Countries with Largest Negative Deviation in Military Expenditure Slope (Cluster 1):")
display(expenditure_slope_deviations.nsmallest(5))

print("\nTop 5 Countries with Largest Positive Deviation in Ongoing Conflict Score Slope (Cluster 1):")
display(conflict_slope_deviations.nlargest(5))

print("\nTop 5 Countries with Largest Negative Deviation in Ongoing Conflict Score Slope (Cluster 1):")
display(conflict_slope_deviations.nsmallest(5))
```

## Conduct granger causality tests

### Subtask:
Perform Granger causality tests to statistically assess if past values of one variable (e.g., military expenditure) predict future values of the other variable (e.g., ongoing conflict), while controlling for their own past values.


**Reasoning**:
Perform Granger causality tests for each country to assess if past values of one differenced variable predict future values of the other differenced variable, using a maximum of 5 lags.



```python
from statsmodels.tsa.stattools import grangercausalitytests

# Define the maximum number of lags for the Granger causality test
max_lag = 5

print("\nGranger Causality Test Results (Military Expenditure_diff -> Ongoing Conflict_diff):")

# Iterate through each country
for country in time_series_df['Country'].unique():
    # Extract differenced columns for the current country, dropping NaNs
    country_data_diff = time_series_df[time_series_df['Country'] == country].dropna(subset=['Military Expenditure_diff', 'Ongoing Conflict_diff']).copy()

    # Ensure there are enough data points for the test
    # Need at least 2 * max_lag + 2 observations
    if len(country_data_diff) >= 2 * max_lag + 2:
        print(f"\n{country}:")
        try:
            # Perform Granger causality test: Military Expenditure_diff -> Ongoing Conflict_diff
            # Null hypothesis: Military Expenditure_diff does NOT Granger-cause Ongoing Conflict_diff
            gc_results_exp_to_conflict = grangercausalitytests(country_data_diff[['Ongoing Conflict_diff', 'Military Expenditure_diff']], max_lag, verbose=False)

            # Print the p-values for each lag
            print("  Military Expenditure_diff Granger-causes Ongoing Conflict_diff (P-values):")
            for lag in range(1, max_lag + 1):
                p_value = gc_results_exp_to_conflict[lag][0]['ssr_ftest'][1]
                print(f"    Lag {lag}: {p_value:.4f}")

        except Exception as e:
            print(f"  Could not perform test: {e}")

    else:
        print(f"\n{country}: Not enough data points ({len(country_data_diff)}) for Granger causality test with max_lag={max_lag}.")

print("\nGranger Causality Test Results (Ongoing Conflict_diff -> Military Expenditure_diff):")

# Iterate through each country again for the reverse test
for country in time_series_df['Country'].unique():
    # Extract differenced columns for the current country, dropping NaNs
    country_data_diff = time_series_df[time_series_df['Country'] == country].dropna(subset=['Military Expenditure_diff', 'Ongoing Conflict_diff']).copy()

    # Ensure there are enough data points for the test
    # Need at least 2 * max_lag + 2 observations
    if len(country_data_diff) >= 2 * max_lag + 2:
        print(f"\n{country}:")
        try:
            # Perform Granger causality test: Ongoing Conflict_diff -> Military Expenditure_diff
            # Null hypothesis: Ongoing Conflict_diff does NOT Granger-cause Military Expenditure_diff
            gc_results_conflict_to_exp = grangercausalitytests(country_data_diff[['Military Expenditure_diff', 'Ongoing Conflict_diff']], max_lag, verbose=False)

            # Print the p-values for each lag
            print("  Ongoing Conflict_diff Granger-causes Military Expenditure_diff (P-values):")
            for lag in range(1, max_lag + 1):
                p_value = gc_results_conflict_to_exp[lag][0]['ssr_ftest'][1]
                print(f"    Lag {lag}: {p_value:.4f}")

        except Exception as e:
            print(f"  Could not perform test: {e}")

    else:
        # Already printed message for insufficient data in the previous loop, so skip.
        pass
```

The previous Granger causality tests failed for all countries because there were not enough data points after differencing (9 points) to perform the test with a maximum lag of 5 (which requires at least 12 points).
