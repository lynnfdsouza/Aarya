## Prepare Data for Clustering

Select the relevant data (GPI scores and domain scores for NATO countries from 2014-2023) and handle missing values for K-means clustering.

## Determine Optimal Number of Clusters for K-Means Clustering

### Subtask:
Use the elbow method to determine the optimal number of clusters for K-means clustering on the prepared NATO countries data.

**Reasoning**:
The elbow method is a common technique to estimate the optimal number of clusters for K-means. It involves fitting the K-means algorithm for a range of cluster numbers and calculating the inertia (sum of squared distances of samples to their closest cluster center) for each number. The optimal number of clusters is often considered to be at the "elbow point" where the rate of decrease in inertia sharply changes. I will calculate the inertia for a range of cluster numbers and plot it to identify the elbow.

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# Select the numerical columns for clustering (excluding 'Country' and 'Year')
X = df_nato_clustering.select_dtypes(include=np.number)

# Scale the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Determine the optimal number of clusters using the Elbow Method
inertia = []
# Trying a range of cluster numbers, for example from 1 to 10
k_range = range(1, 11)

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10) # Added n_init
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

# Plot the elbow method graph
plt.figure(figsize=(10, 6))
plt.plot(k_range, inertia, marker='o')
plt.title('Elbow Method for Optimal K')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.xticks(k_range)
plt.grid(True)
plt.show()
```

## Apply K-Means Clustering

### Subtask:
Apply the K-means clustering algorithm to the prepared and scaled data using the determined number of clusters.

**Reasoning**:
Based on the elbow method, K=3 appears to be a reasonable number of clusters. I will apply the K-means algorithm with 3 clusters to the scaled data. This will assign each data point (representing a country in a specific year) to a cluster.

```python
# Apply K-Means clustering with the determined number of clusters (e.g., K=3)
# Replace 3 with the chosen number of clusters based on the elbow plot if different
optimal_k = 3
kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10) # Added n_init
df_nato_clustering['Cluster'] = kmeans.fit_predict(X_scaled)

# Display the head of the dataframe with the assigned clusters
print("Data with assigned clusters head:")
display(df_nato_clustering.head())

# Get the cluster centers (in scaled space)
cluster_centers_scaled = kmeans.cluster_centers_
print("\nCluster centers (scaled):")
display(cluster_centers_scaled)
```
