## Combine data

### Subtask:
Merge the relevant columns (Overall Score, Safety and Security score, Militarisation score, and Ongoing Conflict score) for NATO countries into a single DataFrame.


**Reasoning**:
Rename the country column, select the relevant columns (Country and years), and merge the filtered dataframes based on the 'Country' column. Then rename the columns in the merged dataframe for clarity.



```python
# Inspect the columns of one of the filtered dataframes to find the exact year column names
print(df_overall_scores_filtered.columns)

# Assuming the year columns are indeed numeric and represent years,
# we will try to select columns that are numeric and within the year range,
# in addition to the 'Country' column.
cols_to_select = ['Country']
year_cols = [col for col in df_overall_scores_filtered.columns if isinstance(col, (int, float)) and col >= 2008 and col <= 2023]
cols_to_select.extend(year_cols)


df_overall_scores_selected = df_overall_scores_filtered[cols_to_select]
df_safety_security_selected = df_safety_security_filtered[cols_to_select]
df_militarisation_selected = df_militarisation_filtered[cols_to_select]
df_ongoing_conflict_selected = df_ongoing_conflict_filtered[cols_to_select]

# Merge the selected columns from the filtered DataFrames
merged_df = df_overall_scores_selected.merge(df_safety_security_selected, on='Country', how='left', suffixes=('_Overall', '_Safety'))
merged_df = merged_df.merge(df_militarisation_selected, on='Country', how='left', suffixes=('_Safety', '_Militarisation'))
merged_df = merged_df.merge(df_ongoing_conflict_selected, on='Country', how='left', suffixes=('_Militarisation', '_Ongoing'))


# Rename the columns in the merged DataFrame
new_columns = {}
for col in merged_df.columns:
    if col != 'Country':
        if isinstance(col, float):
            year = int(col)
            if f'{col}_Overall' in merged_df.columns: # Check if it was an Overall column
                 new_columns[col] = f'{year}_Overall'
            elif f'{col}_Safety' in merged_df.columns: # Check if it was a Safety column
                 new_columns[col] = f'{year}_Safety'
            elif f'{col}_Militarisation' in merged_df.columns: # Check if it was a Militarisation column
                 new_columns[col] = f'{year}_Militarisation'
            elif f'{col}_Ongoing' in merged_df.columns: # Check if it was an Ongoing column
                 new_columns[col] = f'{year}_Ongoing'
            else: # Fallback for columns without explicit suffixes (likely from the first merge or if suffixes were not applied correctly)
                 # We need a more robust way to determine the metric
                 # Let's check if the original selected dataframes have this column to determine the metric
                 if col in df_overall_scores_selected.columns:
                      new_columns[col] = f'{year}_Overall'
                 elif col in df_safety_security_selected.columns:
                      new_columns[col] = f'{year}_Safety'
                 elif col in df_militarisation_selected.columns:
                      new_columns[col] = f'{year}_Militarisation'
                 elif col in df_ongoing_conflict_selected.columns:
                      new_columns[col] = f'{year}_Ongoing'
                 else:
                      # If all checks fail, keep the original column name for inspection
                      new_columns[col] = col

        elif isinstance(col, str) and '_' in col:
             parts = col.split('_')
             if len(parts) > 1 and parts[-1] in ['Overall', 'Safety', 'Militarisation', 'Ongoing']:
                 try:
                     year = int(float(parts[0]))
                     metric = parts[-1]
                     new_columns[col] = f'{year}_{metric}'
                 except ValueError:
                      new_columns[col] = col # Keep original if year is not a valid float
             else:
                 new_columns[col] = col # Keep original if not in expected format
        else:
             new_columns[col] = col # Keep original for other cases (like 'Country')


merged_df = merged_df.rename(columns=new_columns)

# Display the head of the combined DataFrame
display(merged_df.head())
```

```python
# Re-examine the column names and select 'Unnamed: X' columns for years 2008-2023.
# Based on previous output, 'Unnamed: 2' is 2008, 'Unnamed: 3' is 2009, ..., 'Unnamed: 17' is 2023.
year_columns_map = {f'Unnamed: {i}': 2006 + i for i in range(2, 18)}

# Select the 'Country' column and the columns corresponding to the years 2008 to 2023
# Explicitly filter out columns that are not in our expected 'Unnamed: X' format for years.
cols_to_select = ['Country'] + [col for col in df_overall_scores_filtered.columns if col in year_columns_map.keys()]

df_overall_scores_selected = df_overall_scores_filtered[cols_to_select]
df_safety_security_selected = df_safety_security_filtered[cols_to_select]
df_militarisation_selected = df_militarisation_filtered[cols_to_select]
df_ongoing_conflict_selected = df_ongoing_conflict_filtered[cols_to_select]


# Merge the selected columns from the filtered DataFrames
merged_df = df_overall_scores_selected.merge(df_safety_security_selected, on='Country', how='left', suffixes=('_Overall', '_Safety'))
merged_df = merged_df.merge(df_militarisation_selected, on='Country', how='left', suffixes=('_Safety', '_Militarisation'))
merged_df = merged_df.merge(df_ongoing_conflict_selected, on='Country', how='left', suffixes=('_Militarisation', '_Ongoing'))

# Rename the columns in the merged DataFrame
new_columns = {}
for col in merged_df.columns:
    if col != 'Country':
        if '_' in col:
            parts = col.split('_')
            original_col = '_'.join(parts[:-1])
            metric = parts[-1]
            if original_col in year_columns_map:
                year = year_columns_map[original_col]
                new_columns[col] = f'{year}_{metric}'
            else:
                 new_columns[col] = col # Keep original if not in map
        else:
            if col in year_columns_map:
                 year = year_columns_map[col]
                 new_columns[col] = f'{year}_Overall' # Assume columns without suffix are from overall scores
            else:
                 new_columns[col] = col # Keep original if not in map

merged_df = merged_df.rename(columns=new_columns)


# Display the head of the combined DataFrame
display(merged_df.head())
```

## Analyze and compare

### Subtask:
Analyze the combined data to compare NATO countries based on the selected metrics.


**Reasoning**:
Calculate descriptive statistics, average scores, and change over time for each metric across NATO countries and store the results.



```python
import numpy as np

# 1. Calculate descriptive statistics for each metric across all years
metrics = ['Overall', 'Safety', 'Militarisation', 'Ongoing']
descriptive_stats = {}

for metric in metrics:
    # Select columns for the current metric across all years
    metric_cols = [col for col in merged_df.columns if metric in col]

    # Calculate descriptive statistics for these columns
    # Stack the metric columns to treat all years as a single series for overall stats
    stacked_data = merged_df[metric_cols].stack()
    descriptive_stats[metric] = {
        'mean': stacked_data.mean(),
        'median': stacked_data.median(),
        'std': stacked_data.std(),
        'min': stacked_data.min(),
        'max': stacked_data.max()
    }

# Store descriptive statistics in a DataFrame for better presentation
descriptive_stats_df = pd.DataFrame(descriptive_stats).T
display(descriptive_stats_df)

# 2. Calculate the average score for each country across all years for each metric
average_scores_per_country = pd.DataFrame()
average_scores_per_country['Country'] = merged_df['Country']

for metric in metrics:
    metric_cols = [col for col in merged_df.columns if metric in col]
    average_scores_per_country[f'Average_{metric}'] = merged_df[metric_cols].mean(axis=1)

display(average_scores_per_country.head())

# 3. Identify the countries with the lowest and highest average scores for each metric
min_max_average_countries = {}

for metric in metrics:
    min_country = average_scores_per_country.loc[average_scores_per_country[f'Average_{metric}'].idxmin()]
    max_country = average_scores_per_country.loc[average_scores_per_country[f'Average_{metric}'].idxmax()]
    min_max_average_countries[metric] = {
        'lowest_average_country': min_country['Country'],
        'lowest_average_score': min_country[f'Average_{metric}'],
        'highest_average_country': max_country['Country'],
        'highest_average_score': max_country[f'Average_{metric}']
    }

# Store min/max average countries in a DataFrame
min_max_average_countries_df = pd.DataFrame(min_max_average_countries).T
display(min_max_average_countries_df)


# 4. Calculate the change in scores for each metric over the years (2023 vs 2008)
score_change = pd.DataFrame()
score_change['Country'] = merged_df['Country']

for metric in metrics:
    col_2008 = f'2008_{metric}'
    col_2023 = f'2023_{metric}'
    if col_2008 in merged_df.columns and col_2023 in merged_df.columns:
        score_change[f'Change_{metric}_2008_2023'] = merged_df[col_2023] - merged_df[col_2008]
    else:
        print(f"Warning: Could not find columns {col_2008} or {col_2023} for metric {metric}")


# Identify countries with the largest increase and decrease in scores
largest_change_countries = {}

for metric in metrics:
    change_col = f'Change_{metric}_2008_2023'
    if change_col in score_change.columns:
        # Largest increase (most positive change)
        largest_increase_country = score_change.loc[score_change[change_col].idxmax()]
        # Largest decrease (most negative change)
        largest_decrease_country = score_change.loc[score_change[change_col].idxmin()]

        largest_change_countries[metric] = {
            'largest_increase_country': largest_increase_country['Country'],
            'largest_increase_change': largest_increase_country[change_col],
            'largest_decrease_country': largest_decrease_country['Country'],
            'largest_decrease_change': largest_decrease_country[change_col]
        }
    else:
        print(f"Could not calculate largest change for {metric} as change column was not found.")


# Store largest change countries in a DataFrame
largest_change_countries_df = pd.DataFrame(largest_change_countries).T
display(largest_change_countries_df)

# 5. Store results in a dictionary (results_analysis) for easy access
results_analysis = {
    'descriptive_statistics': descriptive_stats_df,
    'average_scores_per_country': average_scores_per_country,
    'min_max_average_countries': min_max_average_countries_df,
    'score_change_2008_2023': score_change,
    'largest_change_countries_2008_2023': largest_change_countries_df
}
```

### Subtask: Analyze trends over time compared to NATO average

**Reasoning**:
Calculate the average scores for all NATO countries for each year for each metric, and then visualize the individual country scores alongside the NATO average to identify trends and outliers.

```python
# Calculate the NATO average for each metric and year
nato_average_trends = pd.DataFrame()

# Get the unique years from the merged_df columns
years = sorted(list(set([int(col.split('_')[0]) for col in merged_df.columns if '_' in col])))
nato_average_trends['Year'] = years

for metric in metrics:
    # Select columns for the current metric across all years and sort them by year
    metric_cols = sorted([col for col in merged_df.columns if metric in col], key=lambda x: int(x.split('_')[0]))

    # Calculate the average for each year across all NATO countries
    # The mean(axis=0) will calculate the average for each column (year)
    nato_average_trends[f'NATO_Average_{metric}'] = merged_df[metric_cols].mean(axis=0).values

# Display the NATO average trends
print("NATO Average Trends (2008-2023):")
display(nato_average_trends)

# 2. Visualize individual country trends vs. NATO average for each metric
for metric in metrics:
    plt.figure(figsize=(14, 7))

    # Plot individual country trends
    for index, row in merged_df.iterrows():
        country = row['Country']
        # Select year columns for the current metric for this country and sort by year
        country_metric_values = row[[col for col in merged_df.columns if metric in col]].values
        plt.plot(years, country_metric_values, marker='o', linestyle='-', label=country, alpha=0.6)

    # Plot NATO average trend
    plt.plot(nato_average_trends['Year'], nato_average_trends[f'NATO_Average_{metric}'], color='black', linewidth=2, label='NATO Average')


    plt.title(f'{metric} Score Trends for NATO Countries vs. NATO Average (2008-2023)')
    plt.xlabel('Year')
    plt.ylabel(f'{metric} Score')
    plt.xticks(years)
    plt.grid(True)
    # Place legend outside the plot
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()
```

**Reasoning**:
Determine the optimal number of clusters for the k-means algorithm using the elbow method, which involves calculating the within-cluster sum of squares for a range of cluster numbers and plotting the results.



```python
from sklearn.cluster import KMeans

# Determine the optimal number of clusters using the Elbow Method
sse = [] # Sum of squared errors
k_range = range(1, 10) # Test k from 1 to 9

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10) # Explicitly set n_init
    kmeans.fit(X_scaled)
    sse.append(kmeans.inertia_)

# Plot the elbow curve
plt.figure(figsize=(8, 5))
plt.plot(k_range, sse, marker='o')
plt.title('Elbow Method to Determine Optimal k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Sum of Squared Errors (SSE)')
plt.xticks(k_range)
plt.grid(True)
plt.show()
```

**Reasoning**:
Apply K-Means clustering to the scaled data using the visually determined optimal number of clusters from the elbow method (let's assume k=3 based on the typical interpretation of an elbow plot where the rate of decrease sharply changes). Then add the cluster labels to the original clustering DataFrame and display the first few rows.



```python
# Apply K-Means clustering with the optimal number of clusters (assuming k=3 from the elbow method)
optimal_k = 3
kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10) # Explicitly set n_init
clusters = kmeans.fit_predict(X_scaled)

# Add the cluster labels to the original df_clustering_data DataFrame
df_clustering_data['Cluster'] = clusters

# Display the first few rows of the DataFrame with cluster labels
print(f"Clustering Results (k={optimal_k}):")
display(df_clustering_data.head())
```
