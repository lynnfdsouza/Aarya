## Perform Exploratory Data Analysis

### Subtask:
Calculate descriptive statistics for the GPI scores across all years and visualize the distribution of GPI scores.

**Reasoning**:
To start the exploratory data analysis, I will calculate descriptive statistics for the numerical columns (the years) to get an overview of the central tendency, dispersion, and shape of the data. Then, I will visualize the distribution of GPI scores for a selected year using a histogram to understand the spread of scores across countries.

```python
# Calculate descriptive statistics for the year columns
display(df_overall.describe())

# Visualize the distribution of GPI scores for the latest year (2023)
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.histplot(df_overall[2023], kde=True)
plt.title('Distribution of GPI Scores in 2023')
plt.xlabel('GPI Score')
plt.ylabel('Frequency')
plt.show()
```

## Analyze GPI Trends Over Time

### Subtask:
Analyze and visualize the trends of the average GPI scores over the years (2008-2023).

**Reasoning**:
To analyze the trends in GPI scores, I will calculate the average GPI score for each year from 2008 to 2023. Then, I will create a line plot to visualize how the average GPI score has changed over this period.

```python
# Calculate the average GPI score for each year
years = range(2008, 2024)
average_gpi_scores = df_overall[years].mean()

# Create a line plot to visualize the trend
plt.figure(figsize=(12, 6))
average_gpi_scores.plot(kind='line', marker='o')
plt.title('Average GPI Score Trend (2008-2023)')
plt.xlabel('Year')
plt.ylabel('Average GPI Score')
plt.xticks(years)
plt.grid(True)
plt.show()
```
